```python
import torch
import gradio as gr
from transformers import pipeline

print("Libraries imported successfully!")
```

```python
generator = pipeline(
    "text-generation",
    model="gpt2"
)

print("Generative AI model loaded successfully!")
```

```python
def generate_story(topic, genre, length):

    prompt = f"Write a {genre} story about {topic}."

    result = generator(
        prompt,
        max_new_tokens=length,
        num_return_sequences=1,
        temperature=0.8,
        do_sample=True,
        top_p=0.9
    )

    return result[0]["generated_text"]
```

```python
demo = gr.Interface(
    fn=generate_story,
    inputs=[
        gr.Textbox(label="Enter Story Topic",
                   placeholder="Example: A robot in space"),

        gr.Dropdown(
            choices=["Adventure", "Horror", "Science Fiction",
                     "Fantasy", "Mystery"],
            label="Select Story Genre"
        ),

        gr.Slider(
            minimum=50,
            maximum=200,
            value=100,
            step=10,
            label="Story Length"
        )
    ],

    outputs=gr.Textbox(label="AI Generated Story",
                       lines=15),

    title="Generative AI – Story Generator",
    description="Generate creative stories using Artificial Intelligence."
)

demo.launch(share=True)
```
