```python
# ==========================================
# AI DEBATE ARENA
# ==========================================

import gradio as gr
import os
from google import genai


# ==========================================
# 🔑 GEMINI CLIENT
# ==========================================

client = genai.Client(
    api_key=os.environ["GEMINI_API_KEY"]
)

# Use the model that we confirmed is working
MODEL = "gemini-2.5-flash"


# ==========================================
# GEMINI FUNCTION
# ==========================================

def ask_gemini(prompt):

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


# ==========================================
# 🟢 SUPPORTER
# ==========================================

def supporter(topic):

    prompt = f"""
You are the SUPPORTER in an AI Debate Arena.

Debate Topic:
{topic}

Personality:
- Confident
- Optimistic
- Persuasive
- Logical

Your job:
1. Argue strongly IN FAVOR of the topic.
2. Give clear reasoning.
3. Give practical examples.
4. Anticipate objections.
5. Defend your position logically.

Do NOT argue against the topic.

Give a clear and convincing debate argument.
"""

    return ask_gemini(prompt)


# ==========================================
# 🔴 OPPOSER
# ==========================================

def opposer(topic, supporter_argument):

    prompt = f"""
You are the OPPOSER in an AI Debate Arena.

Debate Topic:
{topic}

SUPPORTER'S ARGUMENT:
{supporter_argument}

Personality:
- Skeptical
- Critical
- Analytical
- Strong-minded

Your job:
1. Argue strongly AGAINST the topic.
2. Directly challenge the Supporter's argument.
3. Identify weaknesses and assumptions.
4. Give logical counterarguments.
5. Give practical examples.

Do NOT agree with the Supporter.

Give a clear and convincing counterargument.
"""

    return ask_gemini(prompt)


# ==========================================
# 🔵 RESEARCHER
# ==========================================

def researcher(topic, supporter_argument, opposer_argument):

    prompt = f"""
You are the RESEARCHER in an AI Debate Arena.

Debate Topic:
{topic}

SUPPORTER:
{supporter_argument}

OPPOSER:
{opposer_argument}

Personality:
- Neutral
- Analytical
- Evidence-focused
- Logical

Your job:
1. Analyze both arguments.
2. Identify strong reasoning.
3. Identify weak reasoning.
4. Identify factual claims.
5. Identify assumptions.
6. Explain where evidence would be needed.
7. Remain completely neutral.

Do NOT choose a winner.

Clearly separate facts, reasoning, and opinions.
"""

    return ask_gemini(prompt)


# ==========================================
# 🟣 DEVIL'S ADVOCATE
# ==========================================

def devils_advocate(
    topic,
    supporter_argument,
    opposer_argument,
    research
):

    prompt = f"""
You are the DEVIL'S ADVOCATE in an AI Debate Arena.

Debate Topic:
{topic}

SUPPORTER:
{supporter_argument}

OPPOSER:
{opposer_argument}

RESEARCHER:
{research}

Personality:
- Challenging
- Creative
- Skeptical
- Unpredictable

Your job:
1. Challenge BOTH sides.
2. Question their assumptions.
3. Identify weaknesses neither side noticed.
4. Identify contradictions.
5. Introduce alternative perspectives.
6. Ask difficult questions.
7. Explore unusual but logical possibilities.

Do not simply agree with either side.

Your goal is to make the debate deeper and more critical.
"""

    return ask_gemini(prompt)


# ==========================================
# 🟡 JUDGE
# ==========================================

def judge(
    topic,
    supporter_argument,
    opposer_argument,
    research,
    devils_argument
):

    prompt = f"""
You are the JUDGE of an AI Debate Arena.

You are completely neutral.

Debate Topic:
{topic}

========== SUPPORTER ==========
{supporter_argument}

========== OPPOSER ==========
{opposer_argument}

========== RESEARCHER ==========
{research}

========== DEVIL'S ADVOCATE ==========
{devils_argument}

Evaluate the entire debate.

Judge both sides based on:

1. Logic
2. Evidence
3. Relevance
4. Strength of reasoning
5. Response to opposing arguments
6. Overall persuasiveness

Give your final result using this structure:

🏆 WINNER:
[Supporter or Opposer]

🟢 SUPPORTER SCORE:
[number]/10

🔴 OPPOSER SCORE:
[number]/10

💪 SUPPORTER STRENGTHS:
[answer]

⚠️ SUPPORTER WEAKNESSES:
[answer]

💪 OPPOSER STRENGTHS:
[answer]

⚠️ OPPOSER WEAKNESSES:
[answer]

🧠 FINAL REASONING:
[explain why the winner won]

Do not choose a winner based on personal preference.
Judge only the quality of the arguments.
"""

    return ask_gemini(prompt)


# ==========================================
# 🏟️ RUN COMPLETE DEBATE
# ==========================================

def run_debate(topic):

    if not topic or not topic.strip():

        message = "⚠️ Please enter a debate topic."

        return (
            message,
            "",
            "",
            "",
            ""
        )

    try:

        # 🟢 Supporter
        supporter_result = supporter(topic)

        # 🔴 Opposer
        opposer_result = opposer(
            topic,
            supporter_result
        )

        # 🔵 Researcher
        researcher_result = researcher(
            topic,
            supporter_result,
            opposer_result
        )

        # 🟣 Devil's Advocate
        devils_result = devils_advocate(
            topic,
            supporter_result,
            opposer_result,
            researcher_result
        )

        # 🟡 Judge
        judge_result = judge(
            topic,
            supporter_result,
            opposer_result,
            researcher_result,
            devils_result
        )

        return (
            supporter_result,
            opposer_result,
            researcher_result,
            devils_result,
            judge_result
        )

    except Exception as e:

        error_message = f"""
❌ **Debate Error**

`{str(e)}`

Please try again.
"""

        return (
            error_message,
            error_message,
            error_message,
            error_message,
            error_message
        )


# ==========================================
# 🎨 GRADIO INTERFACE
# ==========================================

with gr.Blocks(
    title="AI Debate Arena"
) as demo:

    gr.Markdown(
        """
# 🏟️ AI DEBATE ARENA

### Five AI personalities debate your topic

🟢 **Supporter** → 🔴 **Opposer** → 🔵 **Researcher** → 🟣 **Devil's Advocate** → 🟡 **Judge**
"""
    )

    topic = gr.Textbox(
        label="💬 Enter your debate topic",
        placeholder="Example: Should AI replace human teachers?",
        lines=2
    )

    debate_button = gr.Button(
        "🏟️ START DEBATE",
        variant="primary"
    )

    gr.Markdown("## 🟢 SUPPORTER — FOR the topic")

    supporter_output = gr.Markdown()

    gr.Markdown("## 🔴 OPPOSER — AGAINST the topic")

    opposer_output = gr.Markdown()

    gr.Markdown("## 🔵 RESEARCHER — NEUTRAL ANALYSIS")

    researcher_output = gr.Markdown()

    gr.Markdown("## 🟣 DEVIL'S ADVOCATE — CHALLENGES BOTH")

    devils_output = gr.Markdown()

    gr.Markdown("## 🟡 JUDGE — FINAL VERDICT")

    judge_output = gr.Markdown()


    # ==========================================
    # BUTTON ACTION
    # ==========================================

    debate_button.click(
        fn=run_debate,
        inputs=topic,
        outputs=[
            supporter_output,
            opposer_output,
            researcher_output,
            devils_output,
            judge_output
        ]
    )


# ==========================================
# 🚀 LAUNCH
# ==========================================

demo.launch(
    share=True,
    debug=True
)
```
